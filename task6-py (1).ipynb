```python
from huggingface_hub import login

login(token="YOUR_HF_TOKEN")

print("Hugging Face Login Successful")
```

```python
from transformers import DistilBertTokenizerFast
from transformers import DistilBertForSequenceClassification

model_name = "distilbert-base-cased"

tokenizer = DistilBertTokenizerFast.from_pretrained(model_name)

model = DistilBertForSequenceClassification.from_pretrained(model_name)

print("Model and Tokenizer Loaded Successfully")
```

```python
repo_name = "sanjay-bert-model"

model.push_to_hub(f"sanjay-mlops/{repo_name}")

tokenizer.push_to_hub(f"sanjay-mlops/{repo_name}")

print("Model and Tokenizer Uploaded Successfully")
```

```python
import wandb

wandb.init(project="task6-huggingface")

wandb.run.summary["hf_model_url"] = "https://huggingface.co/sanjay-mlops/sanjay-bert-model"

print("HF Model URL Logged Successfully")
```
